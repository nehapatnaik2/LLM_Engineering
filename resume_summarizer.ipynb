```python
# imports

import os
from dotenv import load_dotenv

from IPython.display import Markdown, display
from openai import OpenAI
from pypdf import PdfReader

# If you get an error running this cell, then please head over to the troubleshooting notebook!
```

```python
# Step 1: Load environment variables and initialize the OpenAI client
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
#client = openai.OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
# Step 2: Read and extract text from the PDF resume
# make sure to change the pdf_path to the path of your resume.
# and thatit is saved in the same directory as this notebook.
pdf_path = "resume.pdf"  
resume_text = ""

try:
    reader = PdfReader(pdf_path)
    # Loop through every page in the PDF and combine the text
    for page in reader.pages:
        text = page.extract_text()
        if text:
            resume_text += text + "\n"
except FileNotFoundError:
    print(f"Error: The file '{pdf_path}' was not found. Please check the file path.")
    exit()
```

```python
# Step 3: Define system and user prompts
system_prompt = (
    "You are an expert HR assistant and technical recruiter. "
    "Provide a professional, well-structured summary of the provided resume. "
    "Highlight key skills, professional experience, and education using clear bullet points."
)
user_prompt = f"Here is the extracted resume text to summarize:\n\n{resume_text}"

# Step 4: Construct the messages payload
messages = [
    {'role': 'system', 'content': system_prompt},
    {'role': 'user', 'content': user_prompt}
]
```

```python
# Step 5: Call the OpenAI API using a valid production model
openai = OpenAI(api_key=api_key)
try:
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages
    )
    # Step 6: Print the generated summary
    print(Markdown(response.choices[0].message.content))
except Exception as e:
    print(f"An API error occurred: {e}")
```
